### Using Context Manager (With)

- It's a good idea to close a file after usage as it will free up the resources
- If we dont close it, garbage collector would close it
- with keyword closes the file as soon as the usage is over

```python
# with
with open('sample1.txt','w') as f:
  f.write('Selmom Bhai')
```

```python
# try f.read()
with open('sample.txt','r') as f:
  print(f.read())
```

```python
# moving within a file --> 10 char then 10 char
with open('sample.txt','r') as f:
  print(f.read(10))
```

```python
with open ('sample.txt','r') as f:
  print(f.read(10))
  print(f.read(10))
```

```python
# benefit? ---> to load a big file in memoery
big_L=['hello world' for i in range(1000)]
with open('big.txt','w') as f:
  f.writelines(big_L)
```

```python
with open('big.txt','r') as f:
  chunk_size=100
  while len(f.read(chunk_size))>0:
    print(f.read(chunk_size),end='')
    f.read(chunk_size)
```

```python
# seek and tell function
with open('sample.txt','r') as f:
  print(f.read(10))
  print(f.tell())
  f.seek(0)
  print(f.read(10))
  print(f.tell())
  # tell batayega ki aap currently kaha pai ho
  # seek bata hai ki kaha ja sakte ho
```

```python
# seek during write
with open('sample.txt','w') as f:
  f.write('hello')
  f.seek(0)
  f.write('X')
```

### Serialization and Deserialization

- **Serialization** - process of converting python data types to JSON format
- **Deserialization** - process of converting JSON to python data types

```python
# serialization using json module
# list
import json
L=[1,2,3,4]
with open('demo.json','w') as f:
  json.dump(L,f)
```

```python
d={
    'name':'hrutuja',
    'age':22,
    'gender':'female'
}
with open('demo.json','w') as f:
  json.dump(d,f,indent=4)
```

```python
# deserialization
import json
with open('demo.json','r') as f:
  print(json.load(f))
```

```python
with open('demo.json','r') as f:
  d=json.load(f)
  print(d)
  print(type(d))
```

```python
# serialize and deserialize tuple
import json
t=(1,2,3,45)
with open('demo.json','w') as f:
  json.dump(t,f)
# a tuple is stored as list and when read the type is list
```

```python
t=(1,2,3,4,5,6)
t[1]=4
print(t)
```

## Serializing and Deserialzing custom objects

```python
class Person:

  def __init__(self,fname,lname,age,gender):
    self.fname=fname
    self.lname=lname
    self.age=age
    self.gender=gender

# format to be stored in the file
# Hrutuja Kalaskar age -> 22 gender ->female
```

```python
person=Person('Hrutuja','Kalaskar',22,'female')
```

```python
import json
with open('demo.json','w') as f:
  json.dump(person,f)
```

```python
def show_object(person):
  if isinstance(person,Person):
    return "{} {} age -> {} gender -> {}".format(person.fname,person.lname,person.age,person.gender)

with open('demo.json','w') as f:
  json.dump(person,f,default=show_object)
```

```python
# desrialize
with open('demo.json','r') as f:
  print(json.load(f))
```

## Pickling
`Pickling` is the process whereby a Python object hierarchy is converted into a byte stream, and `unpickling` is the inverse operation, whereby a byte stream (from a binary file or bytes-like object) is converted back into an object hierarchy.

```python
class Person:
  def __init__(self,name,age):
    self.name=name
    self.age=age

  def display_info(self):
    print('Hi my name is',self.name,'and I am ',self.age,'years old')
```

```python
p=Person('Hrutuja',22)
```

```python
# pickle dump
import pickle
with open('person.pkl','wb') as f:
  pickle.dump(p,f)
```

```python
# pickle load
import pickle
with open('person.pkl','rb') as f:
  p=pickle.load(f)
p.display_info()
```

### Pickle Vs Json

- Pickle lets the user to store data in binary format. JSON lets the user store data in a human-readable text format.
